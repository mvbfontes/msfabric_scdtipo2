```python
from notebookutils import fs
from datetime import datetime

# ============================================================
# CONFIGURAÇÃO
# ============================================================

# Lakehouse padrão anexado ao notebook
lakehouse_path = "/lakehouse/default"

origem = f"{lakehouse_path}/Files/novo"

data_carga = datetime.now().strftime("%Y-%m-%d")
hora_carga = datetime.now().strftime("%H-%M-%S")

destino = f"{lakehouse_path}/Files/processados/{data_carga}/{hora_carga}"

# Arquivos que foram carregados com sucesso
arquivos_processados = [
    "dim_estrutura_vendas.csv",
    "fato_vendas.csv"
]

# ============================================================
# MOSTRA O CONTEXTO
# ============================================================

print(f"Origem:  {origem}")
print(f"Destino: {destino}")

# ============================================================
# CRIA A PASTA DE DESTINO
# ============================================================

fs.mkdirs(destino)

# ============================================================
# MOVE OS ARQUIVOS
# ============================================================

for nome in arquivos_processados:

    origem_arquivo = f"{origem}/{nome}"
    destino_arquivo = f"{destino}/{nome}"

    print(f"\nVerificando: {origem_arquivo}")

    if not fs.exists(origem_arquivo):
        raise Exception(
            f"Arquivo não encontrado: {origem_arquivo}"
        )

    print(f"Movendo: {nome}")

    sucesso = fs.mv(
        origem_arquivo,
        destino_arquivo,
        create_path=True,
        overwrite=True
    )

    if not sucesso:
        raise Exception(
            f"Falha ao mover o arquivo: {nome}"
        )

    print(f"OK: {destino_arquivo}")

print("\n==========================================")
print("Arquivos processados com sucesso.")
print(f"Destino: {destino}")
print("==========================================")
```
